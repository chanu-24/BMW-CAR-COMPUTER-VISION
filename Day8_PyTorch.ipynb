```python
import torch
import torch.nn as nn
import torch.optim as optim

print("🚀 [Day 8] 엔진 ECU 맵핑! 그랜드 학습 루프 가동\n")

# 1. 어제 만든 엔진 가져오기 (가상의 간단한 모델로 대체)
# 입력 차원(3*64*64)을 받아서, 2가지(3시리즈, 5시리즈) 중 하나로 분류!
model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(3 * 64 * 64, 128),
    nn.ReLU(),
    nn.Linear(128, 2)
)

# ----------------------------------------------------
# 💡 [핵심 1] 오차 측정기 (Loss Function) & 스패너 (Optimizer) 세팅
# ----------------------------------------------------
# CrossEntropyLoss: 다중 분류에서 가장 많이 쓰는 배기가스 측정기!
criterion = nn.CrossEntropyLoss()

# SGD 옵티마이저 + Momentum (학습률은 안전하게 0.01로 설정)
# model.parameters()는 엔진 안의 모든 나사(가중치)들을 스패너에 연결하는 겁니다.
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)

# 2. 가짜 데이터 준비 (차량 사진 1장과, 정답지)
dummy_images = torch.randn(1, 3, 64, 64) # 가짜 사진
dummy_labels = torch.tensor([1])         # 정답: 이건 5시리즈(인덱스 1)야!

# ----------------------------------------------------
# 💡 [핵심 2] 전설의 5단계 그랜드 학습 루프! (외워두면 평생 씁니다)
# ----------------------------------------------------
epochs = 5 # 이 과정을 5번 반복하며 튜닝합니다.

for epoch in range(epochs):
    # Step 1: 스패너 초기화 (이전 단계의 찌꺼기 기울기를 0으로 청소)
    optimizer.zero_grad()

    # Step 2: 엔진 시동 (Forward Pass - 예측해봐!)
    outputs = model(dummy_images)

    # Step 3: 배기가스 측정 (Loss 계산 - 정답과 얼마나 틀렸어?)
    loss = criterion(outputs, dummy_labels)

    # Step 4: 원인 분석 (Backward Pass - 역전파로 각 부품의 책임 비율 계산!)
    loss.backward()

    # Step 5: 엔진 튜닝 (Optimizer Step - 계산된 책임만큼 나사 조이기!)
    optimizer.step()

    print(f"🔄 Epoch {epoch+1}/{epochs} 완료 | 현재 오차(Loss): {loss.item():.4f}")

print("\n✅ [Day 8 미션 성공] 오차가 점점 줄어들고 있나요? AI가 스스로 똑똑해지는 순간입니다!")
```

어제 우리는 파이토치로 만든 BMW비전엔진에 가짜 연로를 넣고 시동을 걸어봤다.

하지만 갓 조립한 공장 초기화 상태의 엔진은 3시리즈와 5시리즈를 전혀구분을 못하는 바보상태

오늘은 이 엔진이 정답을 맞힐수 있도록, 과거에 복잡한 파이썬 수식으로 구현하던 SGD,Momentun 같은 최적화 알고리즘들을 파이토리에서 단 한줄로 장착하고 길들이는 것

오늘의 핵심이론: 엔진은 어떻게 스스로 똑똑해질까?

AI가 이미지를 보고 차종을 맞히는 과정은 자동차 정비소에서 엔진을 튜닝하는 과정과 완벽하게 똑같다.

1.시동켜기(Foward): 사진을 보고 AI가 "음..이건 5시리즈 확률 80%!"라고 예측을 뱉어낸다.

2.배기가스 검사(Loss계산):정답지가 "틀렸어! 3시리즈야!"라고 알려주며 ,정답과 AI의 예측이 얼마나 빗나갔는지 오차를 계산

3.원인분석(Backward): "왜 5시리즈라고 착각했지? 아,라디에이터 그릴 필터 가중치가 너무 높게 설정됐구나!하고 역추적(역전파)

4.엔진 세팅 수정(Optimizer step):SGD나 Momentum 같은 최적화 도구를 꺼내서 Loss가 줄어드는 방향으로 내부 가중치W 나사를 살짝 조이거나 푼다,


이 4단계를 수천 번 반복하면 오차가 0에 수렴하는 완벽한 맵핑이 끝난다.

코드를 짜기전에,학습률과 Monentum이 이 '오차의 계곡'
