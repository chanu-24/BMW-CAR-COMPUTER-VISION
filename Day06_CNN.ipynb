```python
import torch
import torch.nn as nn

print("🚀 [Day 6] 고성능 딥러닝 엔진(CNN) 조립을 시작합니다...")

# ----------------------------------------------------
# 💡 [핵심] 나만의 CNN 모델 클래스 만들기 (엔진 설계도)
# ----------------------------------------------------
class BMW_Vision_Engine(nn.Module):
    def __init__(self):
        super(BMW_Vision_Engine, self).__init__()

        # 1번 기통 (First Convolutional Block)
        # in_channels=3 (RGB 컬러 사진), out_channels=16 (16개의 특징 필터 생성)
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)
        self.relu1 = nn.ReLU() # 활성화 함수 (엔진의 스파크 플러그 역할!)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2) # 이미지 크기를 반으로 줄여 연산량 압축

        # 2번 기통 (Second Convolutional Block)
        # 앞에서 나온 16개의 특징을 다시 32개로 더 깊게 분석!
        self.conv2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1)
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)

    def forward(self, x):
        # 설계한 기통(Layer)들을 순서대로 통과하며 이미지가 분석됩니다.
        x = self.conv1(x)
        x = self.relu1(x)
        x = self.pool1(x)

        x = self.conv2(x)
        x = self.relu2(x)
        x = self.pool2(x)
        return x

# ----------------------------------------------------
# 🛠️ 엔진 객체 생성 및 상태 확인
# ----------------------------------------------------
engine = BMW_Vision_Engine()
print("\n✅ 엔진 조립 완료! 설계도를 확인합니다:\n")
print(engine)















```

<인공 뇌세포 CNN 조립>

딥러닝의 꽃이라 불리는 합성곱 신경망은 이 과정을 완전히 자동화한다.

합성곱층(Conv Layer):개발자가 수동으로 필터의 모양을 정해주는 대신,AI가 무작위 숫자로 채워진 수십개의 필터를 이미지 위에 슬라이딩 시킵니다.

스스로 학습:처음엔 노이즈만 뽑아내지만, 수만장의 BMW 사진을 보며 정답을 맞히기 위해 핕터의 가중치들을 스스로 수정해 나간다. 이 거대한 파라미터들을 업데이트 하며 최적의 필터값을 찾아내는 과정에서 SGD,Momentun,RMSProp 같은 최적화 알고리즘들이 완벽한 진가를 발휘하게된다.
