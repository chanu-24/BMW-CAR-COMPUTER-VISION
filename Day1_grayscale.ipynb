```python
import cv2

import numpy as np
from google.colab.patches import cv2_imshow

print("🚀 BMW Vision Project 가동 중...")

# 1. 내가 코랩에 업로드한 파일 이름 (대소문자, 확장자 정확히 일치해야 합니다!)
file_name = 'my_bmw.jpg'

# 2. 로컬 파일 읽어오기 (URL 다운로드 코드가 싹 사라지고 1줄로 끝납니다!)
img = cv2.imread(file_name)

# 혹시 사진 업로드가 안 됐을 경우를 대비한 안전장치
if img is None:
    print(f"❌ 앗! '{file_name}' 파일을 찾을 수 없습니다. 왼쪽 폴더에 잘 올라갔는지 확인해주세요!")
else:
    # 3. 화면에 꽉 차지 않게 사진 크기 줄이기
    img_resized = cv2.resize(img, (600, 400))


print("\n🚗 1. 우리가 보는 BMW (RGB 컬러)")
cv2_imshow(img_resized)


# ----------------------------------------------------
# 💡 [핵심] AI의 연산 속도를 높이기 위한 흑백 변환 (Grayscale)
# ----------------------------------------------------

print("\n🦇 2. AI의 시선으로 본 BMW (흑백 데이터로 경량화)")
gray_img = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray_img)


print("\n✅ [Day 1 미션 성공] AI의 망막에 BMW의 형태가 각인되었습니다!")

```



* 픽셀: 사진을 이루는 아주 작은 네모칸

* RGB채널: 컴퓨터는 색을 표현하기 위해 빨강,초록,파랑 세가지 셀로판지를 겹쳐서 봅니다. 즉, 컬러사진 한장은 사실 3장의 숫자표가 겹쳐진 무거운 데이터입니다.

*자율주행 AI가 1초에 수십 장의 사진을 처리하려면
이 '무거운 컬러 사진'을 가볍게 다루는 법부터 배워야한다.

그 첫걸음이 바로 흑백(Grayscale)변환이다.
색상정보 (RGB 3장)을 날려버리고 명암(1장)만 남겨서
연산 속도를 3배 끌어올리는 것.

인터넷에 있는 BMW G20(3시리즈) 사진을 코랩으로 불러와서
AI가 연산하기 편한 흑백 모드로 변환하는 실무코드이다.
