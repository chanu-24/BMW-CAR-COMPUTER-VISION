```python
import torch
import torch.nn as nn

print("🚀 [Day 6] 고성능 딥러닝 엔진(CNN) 조립을 시작합니다...")

# ----------------------------------------------------
# 💡 [핵심] 나만의 CNN 모델 클래스 만들기 (엔진 설계도)
# ----------------------------------------------------
class BMW_Vision_Engine(nn.Module):
    def __init__(self):
        super(BMW_Vision_Engine, self).__init__()

        # 1번 기통 (First Convolutional Block)
        # in_channels=3 (RGB 컬러 사진), out_channels=16 (16개의 특징 필터 생성)
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)
        self.relu1 = nn.ReLU() # 활성화 함수 (엔진의 스파크 플러그 역할!)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2) # 이미지 크기를 반으로 줄여 연산량 압축

        # 2번 기통 (Second Convolutional Block)
        # 앞에서 나온 16개의 특징을 다시 32개로 더 깊게 분석!
        self.conv2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1)
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)

    def forward(self, x):
        # 설계한 기통(Layer)들을 순서대로 통과하며 이미지가 분석됩니다.
        x = self.conv1(x)
        x = self.relu1(x)
        x = self.pool1(x)

        x = self.conv2(x)
        x = self.relu2(x)
        x = self.pool2(x)
        return x

# ----------------------------------------------------
# 🛠️ 엔진 객체 생성 및 상태 확인
# ----------------------------------------------------
engine = BMW_Vision_Engine()
print("\n✅ 엔진 조립 완료! 설계도를 확인합니다:\n")
print(engine)

print("🚀 [Day 7] 딥러닝 엔진 연료 주입 및 시동 테스트!")


# ----------------------------------------------------
# 💡 [핵심 1] AI용 특수 연료(Tensor) 만들기
# 실전에서는 진짜 이미지를 변환하지만, 오늘은 크기가 (1, 3, 64, 64)인
# 가짜 무작위 데이터(Dummy Data)를 만들어서 엔진 테스트만 해보겠습니다.
# ----------------------------------------------------
# torch.randn: 아무 숫자나 채워 넣은 가짜 텐서 생성
dummy_fuel = torch.randn(1, 3, 64, 64)

print(f"⛽ 주입된 연료(Tensor)의 형태: {dummy_fuel.shape}")
print("   -> (Batch: 1장, Channel: 3개, Height: 64, Width: 64)\n")

# ----------------------------------------------------
# 💡 [핵심 2] 포워드 패스 (Forward Pass, 시동 켜기!)
# 어제 만든 엔진 객체에 연료를 쏙 집어넣습니다.
# ----------------------------------------------------
print("🔥 엔진 점화! (Forward Pass 실행 중...)")


# 이 한 줄이 실행될 때, 어제 정의한 forward() 함수 안의
# conv1 -> relu1 -> pool1 -> conv2 -> relu2 -> pool2 가 순식간에 작동합니다!
output = engine(dummy_fuel)


# ----------------------------------------------------
# 💡 [핵심 3] 배기가스(결과물) 확인하기
# ----------------------------------------------------
print(f"💨 엔진을 통과한 결과물의 형태: {output.shape}")
print("   -> 해석: 1장의 사진이, 32개의 특징(채널)을 가진 16x16 크기의 정보로 압축되었습니다!")
print("\n✅ [Day 7 미션 성공] 엔진 시동 완벽하게 성공! 굉음이 들리는 듯합니다!")
```

어제까지 우리는 이미지를 (세로,가로,색상)의 배열

하지만 파이토치 엔진은 무조건 4차원의 텐서 형태로만 연로로 받아들인다.

그 공식이 바로 자율주행 비전 AI

B,C,H,W

B(Batch Size): 한번에 몇대의 자동차 사진을 엔진에 넣을 것인가?

C(Channel):색상채널수,컬러 사진이면 3(RGB),흑백이면 1

H(Height):사진의 세로 픽셀수(예 64)

W(width):사진의 가로 픽셀수(예 64)

즉, 사진 1장을 엔진에 넣으려면 (1,3,64,64)라는 모양의 덩어리로 포장

이 덩어리가 어제만든 Conv2d 와 Maxpool츠을 통과할떄마다 모양이 어떻게 변하는지 파악
