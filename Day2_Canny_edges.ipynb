```python
import cv2
from google.colab.patches import cv2_imshow

print("🚀 [Day 2] 키드니 그릴 형태 스캔 시작...")

file_name = 'my_bmw.jpg'
img=cv2.imread(file_name)


#크기 줄이기 및 흑백 변환
img_resized=cv2.resize(img,(600,400))
gray_img=cv2.cvtColor(img_resized,cv2.COLOR_BGR2GRAY)

print("\n🦇 1. 흑백으로 변환된 BMW")
cv2_imshow(gray_img)

# 💡 [핵심] Canny Edge 알고리즘으로 윤곽선만 쫙 뽑아내기!
# cv2.Canny(이미지, 최소_임계값, 최대_임계값)

edges=cv2.Canny(gray_img,100,200)

print("\n🦴 2. AI가 스케치한 3시리즈의 뼈대 (Canny Edge)")
cv2_imshow(edges)

print("\n✅ [Day 2 미션 성공] AI가 차의 고유한 디자인 형태를 파악하기 시작했습니다!")
```

-길거리에 파란색 3시리즈와 검은색 3시리즈가 지나간다.
색상은 다르지만 우리는 둘다 3시리즈라는걸안다.
ai도 마찬가지다.
-색상은 도색이나 조명에 따라 변하지만, 차량 고유의 디자인형태는 변하지 않기 때문이다.

-이 형태를 추출하기 위해 ai는 사진에서 픽셀 밝기가 급격하게 변하는 경계선을 찾아낸다, 이것은 윤곽선 추출이라고한다.
